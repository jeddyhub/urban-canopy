## The model

Index block groups by $b \in B$.

**Data**
- $A_b$ - area (m²)
- $s_b$ - current canopy fraction
- $p_b$ - population
- $k_b$ - planting capacity (max trees)
- $c_b$ - cost per tree
- $a$ - canopy added per mature tree (m²)
- $C$ - total budget

**Decision:** $x_b \in \mathbb{Z}_{\ge 0}$, trees planted in $b$.

**Derived:** post-planting canopy fraction $\; f_b(x) = s_b + a x_b / A_b$.

**Constraints**

$$\sum_{b} c_b x_b \le C, \qquad 0 \le x_b \le k_b .$$

**Objective A - utilitarian.** Maximise people-weighted canopy gain:

$$\max \; \sum_b p_b \cdot \frac{a x_b}{A_b}$$

This is linear, so the optimum is a greedy corner solution: pour everything into
the block groups with the highest $p_b / A_b$ until capacity or budget runs out.
Confirm that numerically - it is a good test that your model is wired correctly,
and it is also the reason the utilitarian objective is a bad policy on its own.

**Objective B - Rawlsian max-min.** Raise the floor:

$$\max \; t \quad \text{s.t.} \quad t \le s_b + \frac{a x_b}{A_b} \;\; \forall b$$

Also an LP. Compare the two allocations on a map.

## Sanity checks (do not skip)

  interesting.

- Are any $x_b^\ast$ at their capacity bound? Which?
