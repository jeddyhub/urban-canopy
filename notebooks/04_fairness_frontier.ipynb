## Inequality-averse social welfare

Replace the linear objective with a concave one:

$$\max \; \sum_b p_b \, U_\varepsilon\!\big(f_b(x)\big), \qquad
U_\varepsilon(f) = \begin{cases} \dfrac{f^{1-\varepsilon}}{1-\varepsilon} & \varepsilon \ne 1 \\[4pt] \log f & \varepsilon = 1\end{cases}$$

The inequality-aversion parameter $\varepsilon \ge 0$ interpolates:
$\varepsilon = 0$ is utilitarian, $\varepsilon \to \infty$ approaches max-min.

$U_\varepsilon$ is concave, so on a maximisation problem you can replace it with
its **piecewise-linear upper envelope** and stay in LP territory:

$$\max \sum_b p_b z_b \quad \text{s.t.} \quad z_b \le U(\hat f_j) + U'(\hat f_j)\,(f_b(x) - \hat f_j) \;\; \forall j$$

for breakpoints $\hat f_1 < \dots < \hat f_m$ spanning the plausible range of
canopy fractions. More breakpoints means a tighter approximation and a slower
solve; check how the answer moves as you add them.

## Price of fairness

For each $\varepsilon$, evaluate the *utilitarian* objective at that
allocation and compare to the utilitarian optimum:

$$\mathrm{POF}(\varepsilon) = \frac{W^\ast_{\text{util}} - W_{\text{util}}(x^\ast_\varepsilon)}{W^\ast_{\text{util}}}$$

Plot POF against a measure of the resulting inequality - the Gini coefficient of
canopy fraction, or the minimum canopy fraction achieved. That curve is your
headline figure: it tells a decision-maker exactly what equity costs.

The interesting research claim is not "fairness is expensive" or "fairness is
cheap" - it is *where the curve bends*. If large equity gains are available for
small aggregate loss, that is a strong, quantified argument for the equitable
