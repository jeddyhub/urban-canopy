Suggested sequence:

1. **The problem.** Choropleth of current canopy fraction, with the 15% target
   marked in the legend.
2. **The inequity.** Canopy fraction against median income (or against a heat or
   vulnerability index), with block groups labelled by neighbourhood.
3. **The constraint.** Where planting capacity falls short of what the target
   would require - the places planting alone cannot fix.
4. **The allocations.** Utilitarian vs equity-weighted, side by side on the same
   colour scale.
5. **The trade-off.** The price-of-fairness frontier. This is the money figure.
6. **The recommendation.** One chosen allocation, mapped, with a short table of
   the top ten block groups and what they'd receive.
